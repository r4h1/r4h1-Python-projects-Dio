```python
# Main code to do the text extraction in Portuguese Language
import pytesseract
from PIL import Image

try:
    img = Image.open('Lista de material.jpeg')  # Replace with your image file

    # Preprocessing (optional but often improves results)
    # Convert to grayscale
    img = img.convert('L')

    # Apply thresholding (adjust threshold value as needed)
    threshold = 150  # Adjust as needed
    img = img.point(lambda p: p > threshold and 255)

    # Use pytesseract to extract text
    text = pytesseract.image_to_string(img, lang='por')  # Specify language (Portuguese in this case)
    print(text)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# Code to compare the extracted text to the original text
import pytesseract
from PIL import Image
import difflib

try:
    img = Image.open('Lista de material.jpeg')

    # Preprocessing
    img = img.convert('L')
    threshold = 150
    img = img.point(lambda p: p > threshold and 255)

    # Extract text
    text = pytesseract.image_to_string(img, lang='por')

    # Get the original text (replace with your actual original text)
    with open('original_text.txt', 'r', encoding='utf-8') as f:
        original_text = f.read()

    # Calculate the similarity ratio
    similarity_ratio = difflib.SequenceMatcher(None, original_text, text).ratio()
    accuracy_percentage = similarity_ratio * 100

    print("Extracted Text:")
    print(text)
    print(f"\nAccuracy: {accuracy_percentage:.2f}%")

except Exception as e:
    print(f"An error occurred: {e}")
```
